```python
import numpy as np
import sympy as sp 

x=sp.Symbol('X')
f=x**3-x-1
df=sp.diff(f,x)
F = sp.lambdify(x,f,'numpy')
DF=sp.lambdify(x,df,'numpy')
a,b,h,tol = 1,100,1,0.001

for x in np.arange(a,b,h):
    if F(x) * F(x+h) < 0:
        a = x
        b = x+h
        break 

print("Interval is: ",a,b)
x0 = (a+b)/2
print("Initial approximation",x0)
print(f'{"itr.no":<17}{"x0":<15}{"f(x0)":<12}{"f\'(x0)":<12}{"x1":<13}{"error":<5}')

i = 1
while True:
    x1 = x0 - F(x0)/DF(x0)
    print(f'{i:<17}{x0:<15}{F(x0):<12}{DF(x0):<12}{x1:<13}{abs(x1-x0):<5}')
    if abs(x1-x0)<tol:
        break
    x0=x1
    i+=1
    print(f"\nroot:{x1:<4f}")
```

```python
import math

print("1.Cooling\n2. Growth\n3. Decay")
c = int(input("Choice:"))

if c == 1:
    T0, Ts, t1, T1 = map(float, input("Enter T0 Ts t1 T1:").split())
    k = -math.log((T1-Ts)/(T0-Ts))/t1
    n= int(input("1. Temperature 2. Time 3. Both\nChoice:"))

    if n in (1, 3):
        t = float(input("Time: "))
        print("Temperature =", round(Ts+(T0-Ts)*math.exp(-k*t), 2))

    if n in (2, 3):
        T = float(input("Target temperature:"))
        print("Time =", round(-math.log((T-Ts)/(T0-Ts))/k, 2), "minutes")

elif c in (2, 3):
    N0, t1, N1 = map(float, input("Enter N0 t1 N1 : ").split())
    k = math.log(N1/N0)/t1

    n = int(input("1.Time 2. Amount 3. Both\nChoice: "))
    
    if n in (1, 3):
        t = float(input("Time: "))
        N = N0 * math.exp(k*t)
        print(("Growth" if c == 2 else "Decay"), "=", round(N, 2))

    if n in (2, 3):
        N = float(input("Target amount: "))
        print("Time =", round(math.log(N/N0)/k, 2))
  
else:
    print("Invalid choice") 
    
```

```python

import math

function = input("Enter f(x,y,u): ")

def f(x,y,u):
    return eval(function)

x,y,u = map(float,input("Enter initial values x,y,u: ").split())
h = float(input("Enter step size h: "))
n = int(input("Enter the no.of steps: "))

print(f"{'x':>6}{'y':>12}{'u':>12}")

for i in range(n):

    k1 = h*u
    l1 = h*f(x,y,u)

    k2 = h*(u+l1/2)
    l2 = h*f(x+h/2,y+k1/2,u+l1/2)

    k3 = h*(u+l2/2)
    l3 = h*f(x+h/2,y+k2/2,u+l2/2)

    k4 = h*(u+l3)
    l4 = h*f(x+h,y+k3,u+l3)

    y = y+(k1+2*k2+2*k3+k4)/6
    u = u+(l1+2*l2+2*l3+l4)/6

    x = x+h

    print(f"{x:6.1f}{y:12.6f}{u:12.6f}")
```

```python

import numpy as np
import matplotlib.pyplot as plt

n = int(input("Enter the no.of points: "))

x = np.array(list(map(float,input("Enter x values: ").split())))
y = np.array(list(map(float,input("Enter y values: ").split())))

B,A = np.polyfit(x,np.log(y),1)

a = np.exp(A)
b = B

print(f"\nThe required exponential curve is y = {a:.4f} * e^({b:.4f}x)")

y_fit = a*np.exp(b*x)

plt.scatter(x,y,marker='o',label="Given Data")
plt.plot(x,y_fit,label="Fitted Exponential Curve")

plt.title("Fitting an Exponential Curve")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

n = int(input("Enter the no.of points: "))

x = np.array(list(map(float,input("Enter x values: ").split())))
y = np.array(list(map(float,input("Enter y values: ").split())))

B,A = np.polyfit(np.log(x),np.log(y),1)

a = np.exp(A)
b = B

print(f"\nThe required power curve is y = {a:.4f} * x^{b:.4f}")

y_fit = a*x**b

plt.scatter(x,y,marker='o',label="Given Data")
plt.plot(x,y_fit,label="Fitted Power Curve")

plt.title("Fitting a Power Curve")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```
